# **Variables in Python**

A **variable** is a container used to store data values in computer's physical memory **RAM(Random Access Memory)**. In Python, you do not need to declare the type of variable explicitly unlike SQL; it is created the moment you assign a value to it using the `=` operator.

---

### Variable Naming Rules

| Rule | Allowed / Valid | Not Allowed / Invalid | Reason |
| :--- | :--- | :--- | :--- |
| **Starting Character** | `score`, `_temperature` | `2nd_place` | Must start with a letter or an underscore (`_`), never a number. |
| **Allowed Characters** | `user_score_1` | `user-score`, `user score` | Only letters, digits (`0-9`), and underscores (`_`) are allowed. |
| **Case Sensitivity** | `age`, `Age`, `AGE` | N/A | These are treated as 3 completely different variables. |
| **Reserved Words** | `class_name`, `my_def` | `class`, `def`, `if`, `for` | Reserved Python keywords cannot be used as variable names. |
